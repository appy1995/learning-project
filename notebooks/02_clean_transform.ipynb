```python
from pyspark.sql.functions import col, when

df = spark.read.table("workspace.default.raw_table")
```

```python
df_clean = (
    df
    .dropDuplicates()
    .filter(col("danceability").isNotNull())
    .filter(col("energy").isNotNull())
)

numeric_cols = ["danceability", "energy"]

for c in numeric_cols:
    df_clean = df_clean.withColumn(
        c,
        when(
            col(c).rlike(r"^[-+]?\d*\.?\d+(e[-+]?\d+)?$"),
            col(c).cast("double")
        ).otherwise(None)
    )
```

```python
df_clean.write.mode("overwrite").saveAsTable("workspace.default.clean_table")
```
